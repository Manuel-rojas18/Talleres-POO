# ÁRBOLES DE DECISIÓN EN EL APRENDIZAJE AUTOMÁTICO

## Introducción

Los **árboles de decisión** son uno de los algoritmos más utilizados dentro del aprendizaje automático debido a que permiten representar un proceso de toma de decisiones de una manera sencilla de comprender e interpretar. Su funcionamiento se basa en dividir progresivamente un conjunto de datos utilizando diferentes condiciones hasta llegar a una conclusión o predicción.

A diferencia de algunos modelos matemáticos que pueden ser difíciles de interpretar, un árbol de decisión tiene una estructura similar a un diagrama de flujo. El modelo comienza en un punto inicial denominado **nodo raíz**, realiza diferentes preguntas o condiciones y, dependiendo de las respuestas, continúa por diferentes ramas hasta llegar a un resultado.

Los árboles de decisión pueden utilizarse tanto para **clasificación** como para **regresión**. En clasificación, el objetivo es determinar a qué categoría pertenece un registro. En regresión, el objetivo es estimar un valor numérico, como el precio de una vivienda.

En el contexto de la Ciencia de Datos, este algoritmo es importante porque permite transformar información compleja en una serie de decisiones que pueden ser interpretadas fácilmente. Además, constituye la base de algoritmos más avanzados como **Random Forest** y otros métodos de aprendizaje conjunto.

# 1. ¿Qué es un árbol de decisión?

Un **árbol de decisión** es un modelo de aprendizaje automático supervisado que utiliza una estructura jerárquica de decisiones para realizar predicciones.

El modelo recibe un conjunto de datos de entrenamiento y busca encontrar las características que permiten dividir los registros de manera útil.

Una estructura básica puede representarse así:

**Nodo raíz**

↓

**Pregunta o condición**

↓

**Rama A / Rama B**

↓

**Nuevas condiciones**

↓

**Resultado**

Por ejemplo, si queremos determinar si una persona puede recibir un crédito, el árbol podría realizar preguntas como:

* ¿Tiene ingresos superiores a determinado valor?
* ¿Tiene historial crediticio?
* ¿Cuánto tiempo lleva trabajando?
* ¿Tiene obligaciones pendientes?

Dependiendo de las respuestas, el árbol puede terminar clasificando al cliente en una categoría determinada.

---

# 2. Estructura de un árbol de decisión

Un árbol de decisión está compuesto por diferentes elementos.

## Nodo raíz

Es el punto inicial del árbol.

Representa la primera condición utilizada para dividir los datos.

## Nodos internos

Son los puntos donde el árbol realiza nuevas preguntas o decisiones.

## Ramas

Representan los diferentes caminos que pueden seguir los datos dependiendo de la condición evaluada.

## Nodos hoja

Son los puntos finales del árbol.

Representan el resultado final de la predicción.

Una representación sencilla sería:

```text
                    Área > 100 m²
                    /           \
                  Sí             No
                 /                \
        Habitaciones > 3          Precio bajo
             /      \
           Sí        No
          /            \
    Precio alto      Precio medio
```

En este ejemplo, el árbol utiliza primero el área y posteriormente el número de habitaciones para llegar a una predicción.

---

# 3. Nodo raíz

El nodo raíz es uno de los elementos más importantes porque representa la primera división realizada por el modelo.

El algoritmo analiza las características disponibles y busca aquella condición que permita realizar una separación útil de los datos.

Por ejemplo, si estamos prediciendo el precio de una vivienda, el árbol podría encontrar que una característica relacionada con el tamaño de la vivienda permite separar adecuadamente los registros.

Entonces podría comenzar con una condición como:

**¿Área > 150 m²?**

A partir de esta pregunta, los datos se dividen en diferentes grupos.

---

# 4. Nodos internos

Después de realizar la primera división, el árbol puede continuar dividiendo los grupos resultantes.

Por ejemplo:

**Área > 150 m²**

↓

**Número de habitaciones > 3**

↓

**Antigüedad < 10 años**

Cada una de estas condiciones corresponde a una nueva decisión.

El árbol continúa este proceso hasta alcanzar una condición de parada.

---

# 5. Ramas

Las ramas representan los diferentes caminos que puede seguir un registro.

Por ejemplo:

```text
Área > 150
      |
   ┌──┴──┐
   Sí    No
   |      |
   ↓      ↓
Grupo A  Grupo B
```

Cada vivienda seguirá un camino dependiendo de sus características.

Esto permite que el modelo pueda tomar decisiones diferentes para diferentes tipos de registros.

---

# 6. Nodos hoja

Los nodos hoja representan el resultado final.

En un problema de clasificación, una hoja puede representar una categoría:

* Aprobado.
* Rechazado.

En un problema de regresión, puede representar un valor numérico o una estimación.

Por ejemplo:

**Precio estimado = $250.000**

De esta manera, cada observación termina siguiendo un camino dentro del árbol hasta llegar a una predicción.

# 9. ¿Cómo aprende un árbol de decisión?

El entrenamiento de un árbol consiste en encontrar las divisiones que permitan separar los datos de la manera más útil posible.

El algoritmo analiza las diferentes características y determina dónde realizar las divisiones.

Por ejemplo, puede evaluar:

**Área ≤ 100**

**Área ≤ 150**

**Área ≤ 200**

y determinar cuál de estas divisiones produce una separación más adecuada.

Este proceso se repite en diferentes partes del árbol.

---

# 10. Impureza

Para decidir qué división realizar, los algoritmos de árboles utilizan medidas relacionadas con la **impureza**.

La impureza indica qué tan mezcladas se encuentran las diferentes categorías dentro de un grupo.

En clasificación, algunas medidas utilizadas son:

* Gini.
* Entropía.

Si un grupo contiene registros pertenecientes principalmente a una sola categoría, se considera más puro.

Si contiene muchas categorías mezcladas, presenta mayor impureza.

---

# 11. Índice de Gini

El **índice de Gini** es una medida utilizada frecuentemente para determinar la calidad de una división en árboles de clasificación.

De manera simplificada, permite medir qué tan mezcladas están las clases dentro de un nodo.

Un nodo con menor impureza de Gini presenta una distribución más concentrada en determinadas clases.

El algoritmo compara diferentes divisiones y busca aquellas que permiten reducir la impureza.

---

# 12. Entropía

La **entropía** es otra medida utilizada para evaluar la impureza.

Está relacionada con la cantidad de incertidumbre presente dentro de un conjunto de datos.

Si todos los elementos pertenecen a una misma clase, existe poca incertidumbre.

Si diferentes clases aparecen en proporciones similares, la incertidumbre es mayor.

El algoritmo puede utilizar la reducción de entropía para seleccionar las divisiones.

---

# 13. Ganancia de información

La **ganancia de información** permite medir cuánto mejora una división respecto al estado anterior.

La idea es que una buena división debe reducir la incertidumbre.

De manera conceptual:

**Impureza antes de dividir − Impureza después de dividir = Ganancia**

Una división que genera una mayor reducción de la impureza puede ser seleccionada como una alternativa adecuada.

---

# 14. Árboles de decisión para regresión

En regresión no se busca separar categorías, sino dividir los datos de manera que los valores numéricos dentro de cada grupo sean lo más similares posible.

Por ejemplo, para predecir precios de viviendas, el árbol puede crear grupos de viviendas con características similares.

Una hoja puede terminar representando un valor estimado basado en los datos que llegaron hasta esa hoja.

# 17. Sobreajuste en árboles de decisión

Uno de los principales problemas de los árboles de decisión es el **sobreajuste u overfitting**.

Si el árbol crece demasiado, puede terminar aprendiendo características muy específicas de los datos de entrenamiento.

Por ejemplo, un árbol podría tener una gran cantidad de niveles y realizar muchas divisiones hasta prácticamente memorizar los datos.

Esto puede provocar:

**Excelente rendimiento en entrenamiento**

pero

**Peor rendimiento en datos nuevos.**

Por esta razón es importante controlar la complejidad del árbol.

---

# 19. Número mínimo de muestras

Otro parámetro importante es `min_samples_split`.

Este parámetro establece el número mínimo de registros necesarios para permitir que un nodo se divida.

Por ejemplo:

```python
DecisionTreeRegressor(min_samples_split=20)
```

indica que un nodo debe tener al menos 20 muestras para poder realizar una nueva división.

Esto puede ayudar a controlar la complejidad del modelo.

---

# 20. Poda

La **poda** consiste en reducir partes innecesarias de un árbol para evitar que se vuelva excesivamente complejo.

La idea es conservar las divisiones que realmente aportan información y eliminar aquellas que pueden contribuir al sobreajuste.

La poda puede realizarse durante el entrenamiento o posteriormente, dependiendo del algoritmo y la configuración utilizada.

# 23. Evaluación del árbol

Una vez entrenado el modelo es necesario evaluar su rendimiento.

En regresión pueden utilizarse:

### MAE

El **Error Absoluto Medio** permite conocer cuánto se alejan las predicciones de los valores reales en promedio.

### MSE

El **Error Cuadrático Medio** penaliza especialmente los errores grandes.

### RMSE

La **Raíz del Error Cuadrático Medio** expresa el error nuevamente en las unidades de la variable objetivo.

### R²

El coeficiente **R²** permite analizar qué proporción de la variación de la variable objetivo explica el modelo.

# Conclusión

Los **árboles de decisión** son algoritmos de aprendizaje automático supervisado que permiten resolver problemas de clasificación y regresión mediante una estructura basada en decisiones.

Su funcionamiento se fundamenta en dividir progresivamente los datos utilizando diferentes condiciones. El proceso comienza en un nodo raíz, continúa mediante nodos internos y ramas y termina en nodos hoja que representan las predicciones.

Una de las principales características de los árboles de decisión es su facilidad de interpretación. En lugar de presentar únicamente una ecuación compleja, el modelo puede representar sus decisiones mediante reglas similares a un diagrama de flujo.

Sin embargo, los árboles también presentan riesgos, especialmente el sobreajuste. Un árbol demasiado profundo puede aprender demasiado específicamente los datos de entrenamiento y presentar un rendimiento menor al enfrentarse a datos nuevos. Por esta razón, parámetros como `max_depth`, `min_samples_split` y `min_samples_leaf` son importantes para controlar su complejidad.

Los árboles de decisión pueden evaluarse mediante diferentes métricas dependiendo del problema. Para regresión pueden utilizarse MAE, MSE, RMSE y R², mientras que para clasificación pueden utilizarse accuracy, precision, recall y F1-score.

Además, los árboles de decisión son importantes porque sirven como base para métodos más avanzados como Random Forest, donde múltiples árboles se combinan para obtener una predicción conjunta.

En relación con los temas estudiados anteriormente, los árboles de decisión pueden integrarse dentro de KDD, CRISP-DM y la Ruta de Ciencia de Datos. Principalmente participan durante la etapa de modelado, donde los datos preparados se utilizan para construir modelos capaces de descubrir patrones y realizar predicciones.

En conclusión, el árbol de decisión es una herramienta fundamental para comprender el aprendizaje automático porque permite observar de forma relativamente clara cómo los datos pueden transformarse en reglas, decisiones y predicciones. Su utilidad se extiende desde ejercicios académicos hasta aplicaciones prácticas en áreas como finanzas, comercio, educación, industria, análisis empresarial y predicción de precios de viviendas.
