# **REDES NEURONALES CON KERAS: APRENDIZAJE PROFUNDO Y CONSTRUCCIÓN DE MODELOS**

## Introducción

En el campo de la Ciencia de Datos y el Machine Learning existen problemas que pueden ser difíciles de resolver utilizando algoritmos tradicionales. Cuando los datos presentan relaciones complejas, grandes cantidades de variables o estructuras como imágenes, sonidos y textos, pueden utilizarse modelos más avanzados capaces de aprender representaciones complejas a partir de los datos.

Uno de los modelos más importantes dentro de este campo son las **redes neuronales artificiales**.

Las redes neuronales artificiales están inspiradas conceptualmente en la forma en que las neuronas biológicas procesan información. En Machine Learning, una red neuronal está formada por diferentes unidades matemáticas conectadas entre sí que reciben información, realizan operaciones y producen resultados.

Para construir y entrenar redes neuronales existen diferentes herramientas y bibliotecas. Una de las más conocidas es **Keras**, una API de alto nivel que permite desarrollar modelos de aprendizaje profundo utilizando una sintaxis relativamente sencilla.

La combinación de redes neuronales y Keras permite pasar desde:

**Datos → procesamiento → red neuronal → entrenamiento → evaluación → predicción**

Por esta razón, Keras se convierte en una herramienta importante para comprender y desarrollar modelos de **Deep Learning**.

---

# 1. ¿Qué es una red neuronal artificial?

Una **red neuronal artificial** es un modelo computacional compuesto por unidades llamadas neuronas artificiales, organizadas normalmente en diferentes capas.

Una red básica puede representarse como:

```text
ENTRADA
   ↓
CAPA OCULTA
   ↓
CAPA OCULTA
   ↓
SALIDA
```

Cada neurona recibe información, realiza cálculos matemáticos y transmite un resultado a las siguientes neuronas.

El objetivo de la red es aprender relaciones entre los datos de entrada y el resultado esperado.

---

# 2. Inspiración en las neuronas biológicas

Las redes neuronales artificiales están inspiradas de manera simplificada en el funcionamiento de las neuronas biológicas.

Una neurona biológica recibe señales, las procesa y puede transmitir información.

En una red artificial encontramos una idea similar:

**Entradas → procesamiento → salida**

Sin embargo, es importante aclarar que una red neuronal artificial es un modelo matemático y computacional; no representa literalmente el funcionamiento completo del cerebro humano.

---

# 3. La neurona artificial

Una neurona artificial recibe diferentes valores de entrada.

Por ejemplo:

```text
x₁
x₂
x₃
```

Cada entrada tiene asociado un peso:

```text
w₁
w₂
w₃
```

La neurona calcula una combinación de estas entradas.

Una representación simplificada es:

**z = x₁w₁ + x₂w₂ + x₃w₃ + b**

Donde:

* `x` representa las entradas.
* `w` representa los pesos.
* `b` representa el sesgo o bias.
* `z` representa el resultado antes de aplicar la función de activación.

Posteriormente se aplica una función de activación.

---

# 4. Pesos

Los **pesos** son valores que determinan la importancia relativa de las entradas para una neurona.

Durante el entrenamiento, la red modifica estos pesos.

Por ejemplo, inicialmente podemos tener:

**w₁ = 0.2**

**w₂ = -0.5**

**w₃ = 0.7**

Después del entrenamiento pueden cambiar.

El objetivo es encontrar valores que permitan que la red produzca predicciones cada vez más adecuadas.

---

# 5. Bias o sesgo

El **bias** es otro parámetro que permite desplazar la función aprendida por una neurona.

La expresión general puede representarse como:

**z = Σ(xᵢwᵢ) + b**

El modelo aprende tanto los pesos como los valores de bias durante el entrenamiento.

---

# 6. Funciones de activación

Después de realizar la combinación de entradas y pesos, la neurona utiliza una **función de activación**.

Las funciones de activación permiten introducir comportamientos no lineales en la red.

Entre las más conocidas encontramos:

* ReLU.
* Sigmoid.
* Tanh.
* Softmax.

---

# 7. Función ReLU

La función **ReLU** significa:

**Rectified Linear Unit**

Su expresión es:

**ReLU(x) = max(0, x)**

Esto significa que:

* Si `x` es negativo → devuelve 0.
* Si `x` es positivo → devuelve `x`.

ReLU es muy utilizada en capas ocultas de redes neuronales.

---

# 8. Función Sigmoid

La función sigmoid transforma los valores a un rango entre 0 y 1.

Su comportamiento permite utilizarla en determinados problemas de clasificación binaria.

Por ejemplo:

**0.05 → baja probabilidad**

**0.90 → alta probabilidad**

Sin embargo, una salida sigmoid debe interpretarse de acuerdo con cómo fue diseñado el modelo y el problema.

---

# 9. Función Softmax

Softmax se utiliza frecuentemente cuando tenemos clasificación multiclase.

Supongamos que queremos clasificar una imagen en:

* Gato.
* Perro.
* Caballo.

La red puede producir probabilidades aproximadas:

**Gato → 0.10**

**Perro → 0.75**

**Caballo → 0.15**

La suma de las probabilidades es:

**1.00**

La clase con mayor probabilidad puede utilizarse como predicción del modelo.

# 14. ¿Qué es Keras?

**Keras** es una API de alto nivel utilizada para construir y entrenar modelos de aprendizaje profundo.

Permite definir redes neuronales utilizando componentes relativamente sencillos.

Un modelo básico puede construirse utilizando:

```python
model = Sequential()
```

y posteriormente agregando capas.

La idea principal es facilitar el desarrollo de modelos sin tener que implementar manualmente todas las operaciones matemáticas de una red neuronal.

---

# 15. Modelo Sequential

Uno de los enfoques más sencillos de Keras es `Sequential`.

Permite construir una red agregando capas una después de otra.

Por ejemplo:

```python
from tensorflow import keras

model = keras.Sequential([
    keras.layers.Dense(64, activation="relu"),
    keras.layers.Dense(32, activation="relu"),
    keras.layers.Dense(1)
])
```

La estructura sería:

```text
Entrada
   ↓
64 neuronas + ReLU
   ↓
32 neuronas + ReLU
   ↓
1 neurona
```

---

# 16. La capa Dense

`Dense` representa una capa completamente conectada.

Por ejemplo:

```python
keras.layers.Dense(64, activation="relu")
```

significa que tenemos una capa con:

**64 neuronas**

y utilizamos:

**ReLU**

como función de activación.

Cada neurona de una capa Dense está conectada con las salidas de la capa anterior.

# 17. ¿Qué significa Deep Learning?

**Deep Learning** o aprendizaje profundo es una rama del Machine Learning que utiliza redes neuronales con múltiples capas para aprender representaciones complejas.

La palabra **deep** hace referencia a la profundidad de la red.

Una estructura sencilla podría tener:

```text
Entrada
↓
Capa 1
↓
Capa 2
↓
Capa 3
↓
Salida
```

Una red más profunda puede contener muchas más capas.

# 18. Entrenamiento de una red neuronal

El entrenamiento consiste en permitir que la red aprenda los parámetros que necesita para realizar predicciones.

El proceso general es:

**Datos**

↓

**Predicción**

↓

**Cálculo del error**

↓

**Retropropagación**

↓

**Actualización de pesos**

↓

**Nueva predicción**

↓

**Repetición**

Este proceso ocurre muchas veces durante el entrenamiento.

---

# 19. Predicción inicial

Al comenzar el entrenamiento, los pesos de la red normalmente no producen buenas predicciones.

La red realiza una primera predicción.

Por ejemplo:

**Precio real = 300 millones**

**Predicción = 220 millones**

Existe una diferencia entre ambos valores.

Esta diferencia se utiliza para calcular la pérdida.

---

# 20. Función de pérdida

La **función de pérdida** mide qué tan diferente es la predicción respecto al resultado esperado.

En regresión puede utilizarse:

**Mean Squared Error (MSE)**

Una expresión simplificada es:

**MSE = promedio((valor real − predicción)²)**

Mientras más grande sea el error, mayor será la pérdida.

Durante el entrenamiento se intenta reducir esta pérdida.

---

# 21. Optimización

El optimizador determina cómo se actualizan los pesos de la red para reducir la función de pérdida.

Uno de los optimizadores más utilizados es:

**Adam**

En Keras podemos configurarlo:

```python
optimizer="adam"
```

Adam utiliza información de los gradientes para ajustar los parámetros del modelo durante el entrenamiento.

---

# 22. Gradiente

El gradiente proporciona información sobre cómo cambia la función de pérdida cuando cambian los parámetros del modelo.

De manera conceptual:

**Gradiente → dirección de cambio**

El optimizador utiliza esta información para decidir cómo modificar los pesos.

Esto está relacionado con el concepto de **descenso del gradiente**.

---

# 23. Backpropagation

La **retropropagación**, o **backpropagation**, es un procedimiento fundamental para entrenar redes neuronales.

Después de realizar una predicción:

1. Se calcula la pérdida.
2. Se calcula cómo contribuyeron los parámetros al error.
3. Se propagan los gradientes hacia atrás.
4. Se actualizan los pesos.

Podemos representarlo:

```text
Entrada
  ↓
Predicción
  ↓
Pérdida
  ↓
Gradientes ←
  ↓
Actualización
```

Este proceso se repite durante el entrenamiento.

---

# 24. Epoch

Una **epoch** representa una pasada completa por el conjunto de datos de entrenamiento.

Por ejemplo:

```python
model.fit(X_train, y_train, epochs=50)
```

significa que el modelo realizará hasta 50 pasadas sobre los datos de entrenamiento, dependiendo de otras configuraciones.

Una mayor cantidad de epochs no garantiza un mejor modelo.

Si entrenamos demasiado, podemos producir **overfitting**.

---

# 25. Batch

Un **batch** es un subconjunto de los datos utilizado para realizar una actualización durante el entrenamiento.

Por ejemplo, si tenemos:

**10.000 registros**

y:

**batch_size = 32**

el modelo procesa los datos en grupos de 32 aproximadamente.

Esto permite trabajar con grandes conjuntos de datos sin tener que procesarlos todos simultáneamente en cada actualización.

---

# 26. Entrenamiento y validación

Es importante separar los datos.

Podemos tener:

**Datos originales**

↓

**Entrenamiento**

**Validación**

**Prueba**

Los datos de entrenamiento se utilizan para aprender.

Los datos de validación ayudan a controlar el comportamiento del modelo durante el desarrollo.

Los datos de prueba permiten realizar una evaluación final utilizando datos que el modelo no utilizó para ajustar sus parámetros.

---

# 27. Overfitting

El **overfitting** ocurre cuando el modelo aprende demasiado bien los datos de entrenamiento, incluyendo patrones específicos o ruido que no se generalizan correctamente.

Podemos observar:

**Entrenamiento → excelente**

**Datos nuevos → peor rendimiento**

Esto indica que el modelo puede estar memorizando características específicas de los datos de entrenamiento.

---

# 28. Underfitting

El **underfitting** ocurre cuando el modelo es demasiado simple para representar adecuadamente las relaciones presentes en los datos.

Podemos observar:

**Entrenamiento → rendimiento bajo**

**Validación → rendimiento bajo**

En este caso puede ser necesario revisar:

* Arquitectura.
* Variables.
* Entrenamiento.
* Hiperparámetros.
* Calidad de los datos.

---

# 29. Regularización

La regularización busca reducir el riesgo de overfitting.

Algunas técnicas incluyen:

* L1.
* L2.
* Dropout.
* Early Stopping.

Estas técnicas introducen restricciones o estrategias que pueden ayudar a mejorar la generalización.

---

# 30. Dropout

**Dropout** consiste en desactivar aleatoriamente una proporción de neuronas durante el entrenamiento.

Por ejemplo:

```python
keras.layers.Dropout(0.2)
```

significa que durante el entrenamiento se puede omitir aproximadamente el 20% de las activaciones de esa capa en cada actualización.

El objetivo es evitar que la red dependa demasiado de determinadas conexiones.

---

# 31. Early Stopping

Otra técnica es **Early Stopping**.

La idea consiste en detener el entrenamiento cuando el rendimiento de validación deja de mejorar.

Por ejemplo:

```python
early_stop = keras.callbacks.EarlyStopping(
    patience=5,
    restore_best_weights=True
)
```

Esto puede ayudar a evitar entrenar durante demasiadas epochs cuando el modelo ya no está mejorando.

# 59. Ejemplo conceptual completo

Supongamos que queremos predecir el precio de una vivienda.

Tenemos:

* Área.
* Habitaciones.
* Baños.
* Año de construcción.
* Garaje.

Primero obtenemos los datos.

Después limpiamos los registros.

Posteriormente dividimos:

**80% → entrenamiento**

**20% → validación**

Construimos la red:

```text
5 variables
    ↓
64 neuronas
    ↓
32 neuronas
    ↓
1 neurona
    ↓
Precio
```

La red comienza con pesos iniciales.

Realiza predicciones.

Calculamos el error.

La retropropagación calcula gradientes.

El optimizador actualiza los pesos.

El proceso continúa durante varias epochs.

Finalmente evaluamos el modelo con datos que no fueron utilizados para entrenarlo.

# 60. Conclusión

Las **redes neuronales artificiales** constituyen una de las herramientas más importantes dentro del Machine Learning y el Deep Learning. Su capacidad para aprender relaciones complejas a partir de ejemplos permite abordar problemas que pueden resultar difíciles para modelos más sencillos.

**Keras** facilita la construcción de estas redes mediante una API de alto nivel que permite definir capas, funciones de activación, optimizadores, funciones de pérdida y métricas sin tener que implementar manualmente todas las operaciones matemáticas necesarias.

El funcionamiento de una red neuronal se basa en elementos fundamentales como **neuronas, pesos, bias, funciones de activación, capas, pérdida, optimización y retropropagación**. Durante el entrenamiento, el modelo modifica sus parámetros para reducir el error y mejorar sus predicciones.

Conceptos como **epochs, batches, learning rate, overfitting, underfitting, regularización y dropout** permiten comprender cómo controlar el proceso de aprendizaje y mejorar la capacidad de generalización del modelo.

Keras también permite trabajar con diferentes tipos de problemas. En regresión puede utilizarse para predecir valores continuos, mientras que en clasificación puede utilizarse para identificar categorías. Además, mediante arquitecturas como las CNN, puede utilizarse para trabajar con imágenes y otros tipos de datos complejos.

La relación con conceptos estudiados anteriormente también es importante. Las redes neuronales pueden formar parte de procesos de **Ciencia de Datos, KDD y CRISP-DM**, especialmente durante las etapas de modelado y evaluación.

Sin embargo, construir una red neuronal no consiste únicamente en agregar muchas capas y neuronas. Un buen proyecto requiere comprender el problema, preparar correctamente los datos, seleccionar una arquitectura apropiada, evaluar el modelo y analizar sus errores.

En conclusión, las redes neuronales con Keras permiten transformar datos en modelos capaces de aprender patrones y realizar predicciones. Su importancia radica no solamente en la potencia matemática de sus algoritmos, sino en la posibilidad de integrarlas dentro de un proceso completo de Ciencia de Datos para convertir información en conocimiento y apoyar la solución de problemas reales.
