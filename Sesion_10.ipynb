# Vectores de soporte y máquinas de vectores de soporte (SVM)

## Introducción

Dentro del aprendizaje automático existen diferentes algoritmos diseñados para encontrar patrones en los datos y utilizarlos posteriormente para realizar predicciones. Uno de estos algoritmos es la **Máquina de Vectores de Soporte**, conocida como **Support Vector Machine o SVM**.

Las máquinas de vectores de soporte son algoritmos de aprendizaje supervisado que pueden utilizarse principalmente para resolver problemas de **clasificación**, aunque también existen versiones utilizadas para problemas de regresión.

La idea fundamental de SVM consiste en encontrar una frontera que permita separar diferentes grupos de datos de la mejor manera posible. Para conseguirlo, el algoritmo busca establecer un límite que mantenga la mayor distancia posible respecto a los datos más cercanos de cada grupo.

Los puntos de datos que se encuentran más cerca de esta frontera reciben el nombre de **vectores de soporte**, ya que tienen una influencia fundamental en la posición de la frontera de decisión.

Este concepto permite comprender cómo un algoritmo puede utilizar la geometría y las matemáticas para clasificar información y posteriormente realizar predicciones sobre nuevos datos.

---

# 1. ¿Qué es una Máquina de Vectores de Soporte?

Una **Máquina de Vectores de Soporte (SVM)** es un algoritmo de aprendizaje automático supervisado que busca encontrar una frontera óptima para separar diferentes clases de datos.

Por ejemplo, imaginemos que tenemos información sobre dos tipos de viviendas:

* Viviendas de precio bajo.
* Viviendas de precio alto.

Cada vivienda puede representarse mediante características como:

* Área.
* Número de habitaciones.
* Número de baños.
* Antigüedad.

Si representamos los datos gráficamente, es posible que las viviendas pertenecientes a cada grupo aparezcan en diferentes regiones.

SVM intenta encontrar una frontera que permita separar ambos grupos.

En un problema sencillo de dos dimensiones, esta frontera puede ser una línea.

En dimensiones superiores, puede convertirse en un **hiperplano**.

---

# 2. ¿Qué es un vector?

Para comprender SVM es necesario entender primero el concepto de vector.

En Machine Learning, un registro puede representarse mediante un conjunto de valores numéricos.

Por ejemplo, una vivienda podría representarse como:

**X = [120, 3, 2]**

Donde:

* 120 representa el área.
* 3 representa el número de habitaciones.
* 2 representa el número de baños.

Este conjunto de valores puede representarse matemáticamente como un **vector**.

Por lo tanto, cada observación de un conjunto de datos puede convertirse en un vector de características.

---

# 3. ¿Qué son los vectores de soporte?

Los **vectores de soporte** son los puntos de datos que se encuentran más cerca de la frontera de decisión encontrada por el algoritmo.

Estos puntos son especialmente importantes porque ayudan a determinar la posición y orientación de dicha frontera.

Por ejemplo, imaginemos dos grupos de puntos:

**Grupo A:** ● ● ● ●

**Grupo B:** ▲ ▲ ▲ ▲

Entre ambos grupos existe una frontera.

Los puntos que están más cerca de esa frontera son los vectores de soporte.

Una característica importante de SVM es que no todos los datos tienen necesariamente la misma influencia sobre la frontera. Los puntos más cercanos tienen un papel fundamental.

---

# 4. ¿Qué es un hiperplano?

El **hiperplano** es la frontera que utiliza SVM para separar las diferentes clases.

En dos dimensiones podemos imaginarlo como una línea:

**w₁x₁ + w₂x₂ + b = 0**

Donde:

* **x₁ y x₂** son las características.
* **w₁ y w₂** son los pesos.
* **b** representa el término independiente.

En tres dimensiones, la frontera puede ser un plano.

Cuando existen más dimensiones, hablamos de un hiperplano.

Por esta razón, aunque un conjunto de datos pueda tener muchas características, SVM puede trabajar buscando una frontera matemática que permita separar las clases.

---

# 5. El margen

Uno de los conceptos más importantes de SVM es el **margen**.

El margen representa la distancia entre la frontera de decisión y los vectores de soporte.

La idea principal es encontrar una frontera que no solamente separe las clases, sino que mantenga la mayor distancia posible respecto a los datos más cercanos.

De manera simplificada:

**Clase A ← margen → frontera ← margen → Clase B**

El algoritmo intenta maximizar esta separación.

Por eso se habla de un **margen máximo**.

---

# 6. ¿Por qué se busca maximizar el margen?

Maximizar el margen puede ayudar a obtener una frontera que generalice mejor frente a nuevos datos.

Supongamos que tenemos dos posibles fronteras:

### Frontera A

Separa los datos, pero está muy cerca de algunos puntos.

### Frontera B

También separa los datos, pero mantiene una distancia mayor respecto a los puntos más cercanos.

SVM busca una solución basada en la idea de maximizar el margen.

Esto permite reducir la sensibilidad de la frontera frente a pequeñas variaciones de los datos.

---

# 7. Clasificación con SVM

SVM se utiliza principalmente en problemas de clasificación.

Un problema de clasificación consiste en asignar una observación a una categoría.

Por ejemplo:

* Correo deseado / correo no deseado.
* Producto defectuoso / producto no defectuoso.
* Clase A / Clase B.
* Imagen de gato / imagen de perro.
* Cliente perteneciente a un grupo / cliente perteneciente a otro grupo.

El modelo recibe datos de entrenamiento donde las clases son conocidas y aprende una frontera que posteriormente puede utilizar para clasificar nuevos registros.

---

# 8. Ejemplo sencillo

Supongamos que tenemos estudiantes y queremos clasificarlos según dos características:

* Horas de estudio.
* Calificación obtenida.

Podemos tener dos categorías:

**Clase 0:** estudiantes que no aprobaron.

**Clase 1:** estudiantes que aprobaron.

El algoritmo SVM analiza los datos de entrenamiento y busca una frontera que separe ambas clases.

Posteriormente, si aparece un nuevo estudiante, el modelo puede analizar su posición respecto a la frontera y asignarlo a una de las categorías.

---

# 9. SVM para clasificación lineal

Cuando los datos pueden separarse mediante una línea o hiperplano, podemos utilizar un **SVM lineal**.

Por ejemplo:

**w · x + b = 0**

La frontera divide el espacio en diferentes regiones.

Los datos ubicados a un lado pertenecen a una clase y los datos ubicados al otro lado pertenecen a otra clase.

Este es uno de los casos más sencillos de utilizar SVM.

---

# 10. Cuando los datos no son linealmente separables

En problemas reales, los datos no siempre pueden separarse mediante una línea recta.

Por ejemplo, podemos encontrar una distribución como:

* Una clase ubicada en el centro.
* Otra clase formando un círculo alrededor.

En este caso, una línea recta no sería suficiente para separar correctamente las clases.

Aquí aparecen herramientas más avanzadas de SVM, especialmente los **kernels**.

---

# 11. ¿Qué es un kernel?

Un **kernel** es una función que permite trabajar con relaciones más complejas entre los datos.

La idea es transformar matemáticamente la representación de los datos para que un problema que inicialmente no puede separarse linealmente pueda ser tratado mediante una frontera más apropiada.

Algunos kernels conocidos son:

* Linear.
* Polynomial.
* RBF.
* Sigmoid.

El kernel **RBF**, o Radial Basis Function, es uno de los utilizados frecuentemente cuando los datos presentan relaciones no lineales.

---

# 12. Kernel lineal

El kernel lineal se utiliza cuando existe una relación que puede representarse adecuadamente mediante una separación lineal.

Conceptualmente, busca una frontera de la forma:

**w · x + b = 0**

Es una alternativa sencilla y puede ser eficiente cuando existe una separación aproximadamente lineal.

---

# 13. Kernel polinomial

El kernel polinomial permite representar relaciones más complejas utilizando funciones polinómicas.

Esto puede ser útil cuando la separación entre las clases presenta una estructura curva.

La complejidad dependerá, entre otros aspectos, del grado del polinomio utilizado.

---

# 14. Kernel RBF

El kernel RBF permite trabajar con relaciones no lineales y puede generar fronteras de decisión complejas.

Una de sus características es que considera la distancia entre diferentes puntos dentro del espacio de características.

Este kernel es bastante utilizado en aplicaciones donde los datos no presentan una separación lineal sencilla.

---

# 15. Parámetro C

Uno de los parámetros importantes de SVM es **C**.

El parámetro C controla el equilibrio entre:

* Obtener una frontera con un margen amplio.
* Penalizar los errores de clasificación.

Un valor de C más alto puede hacer que el modelo penalice con mayor fuerza los errores de entrenamiento.

Un valor menor permite una mayor tolerancia a algunos errores a cambio de buscar una frontera con mayor margen.

Por esta razón, C debe seleccionarse cuidadosamente.

---

# 16. Parámetro gamma

Cuando se utilizan determinados kernels, como RBF, aparece otro parámetro importante denominado **gamma**.

Gamma determina qué tan lejos llega la influencia de un solo punto de entrenamiento.

De manera conceptual:

* Gamma bajo → influencia más amplia.
* Gamma alto → influencia más localizada.

Un valor demasiado alto puede producir un modelo excesivamente complejo y aumentar el riesgo de sobreajuste.

---

# 17. SVM y sobreajuste

Al igual que otros algoritmos de Machine Learning, SVM puede presentar **sobreajuste**.

El sobreajuste ocurre cuando el modelo aprende demasiado específicamente los datos de entrenamiento y no consigue generalizar correctamente frente a datos nuevos.

Esto puede suceder especialmente cuando se utilizan configuraciones que generan fronteras demasiado complejas.

Por esta razón es importante evaluar el modelo utilizando datos que no hayan sido utilizados directamente durante el entrenamiento.

---

# 18. SVM y regresión

Aunque SVM es especialmente conocida por la clasificación, también puede utilizarse para regresión.

Esta variante se conoce como:

**Support Vector Regression (SVR)**

En lugar de buscar separar clases, SVR intenta encontrar una función que represente adecuadamente una variable numérica.

Por ejemplo, podría utilizarse para:

* Predecir precios de viviendas.
* Estimar temperaturas.
* Predecir ventas.
* Estimar demanda.

Por lo tanto:

**SVM → principalmente clasificación**

**SVR → regresión**

---

# 19. Diferencia entre SVM y SVR

| Característica   | SVM                    | SVR                  |
| ---------------- | ---------------------- | -------------------- |
| Tipo de problema | Clasificación          | Regresión            |
| Resultado        | Categoría              | Valor numérico       |
| Ejemplo          | Aprobado / No aprobado | Precio               |
| Objetivo         | Separar clases         | Estimar un valor     |
| Frontera         | Hiperplano             | Función de regresión |

Aunque ambos pertenecen a la misma familia de métodos, su objetivo es diferente.

---

# 20. Importancia de la escala de los datos

SVM utiliza distancias y relaciones geométricas entre los datos. Por esta razón, la escala de las variables puede influir considerablemente en su funcionamiento.

Supongamos que tenemos:

**Edad:** 18–70

**Salario:** 1.000–10.000.000

La diferencia de escala es muy grande.

Una variable podría terminar teniendo una influencia desproporcionada.

Por esta razón, generalmente se recomienda aplicar técnicas como:

* Estandarización.
* Normalización.

Una herramienta frecuente es `StandardScaler` de Scikit-learn.

# 22. Evaluación de un modelo SVM

Después del entrenamiento es necesario evaluar el modelo.

Para clasificación pueden utilizarse métricas como:

* Accuracy.
* Precision.
* Recall.
* F1-score.
* Matriz de confusión.

### Accuracy

Representa la proporción de predicciones correctas respecto al total de predicciones.

### Precision

Permite analizar qué proporción de las predicciones positivas realizadas por el modelo fueron realmente positivas.

### Recall

Permite analizar qué proporción de los casos positivos reales fueron identificados correctamente.

### F1-score

Combina precision y recall en una sola métrica.

---

# 23. Matriz de confusión

La matriz de confusión permite observar los resultados de clasificación de una manera más detallada.

En un problema binario pueden aparecer:

* Verdaderos positivos.
* Verdaderos negativos.
* Falsos positivos.
* Falsos negativos.

Esto permite comprender no solamente cuántas predicciones fueron correctas, sino también qué tipo de errores cometió el modelo.

# 24. Ventajas de las máquinas de vectores de soporte

Entre las características que pueden hacer interesante a SVM se encuentran:

* Puede funcionar bien con conjuntos de datos de tamaño pequeño o mediano.
* Puede trabajar con espacios de características de alta dimensión.
* Utiliza los vectores de soporte para definir la frontera.
* Puede resolver problemas lineales y no lineales.
* Los kernels permiten representar relaciones complejas.
* Puede utilizarse tanto para clasificación como para regresión.

---

# 25. Desventajas de SVM

También existen algunas limitaciones.

* Puede requerir una adecuada selección de parámetros.
* La elección del kernel puede afectar considerablemente los resultados.
* Es sensible a la escala de las variables.
* Puede ser costosa computacionalmente en conjuntos de datos muy grandes.
* La interpretación de algunos modelos con kernels no lineales puede ser más complicada.
* Una configuración inadecuada puede provocar sobreajuste.

Por esta razón, SVM debe evaluarse según las características del problema y del conjunto de datos.

# Conclusión

Las **Máquinas de Vectores de Soporte (SVM)** son algoritmos de aprendizaje automático supervisado que buscan encontrar una frontera óptima para separar diferentes categorías de datos.

Su funcionamiento se basa en conceptos como vectores, hiperplanos, márgenes y vectores de soporte. Los vectores de soporte son especialmente importantes porque corresponden a los datos que se encuentran más cerca de la frontera de decisión y contribuyen de manera significativa a determinar su posición.

Una de las características más importantes de SVM es su capacidad para trabajar tanto con problemas lineales como con problemas no lineales mediante el uso de diferentes funciones kernel. Entre ellas se encuentran los kernels lineal, polinomial y RBF.

Además, existen parámetros como **C** y **gamma** que permiten controlar diferentes aspectos del comportamiento del modelo. Una configuración adecuada puede ayudar a obtener una buena capacidad de generalización, mientras que una configuración inadecuada puede producir problemas como el sobreajuste.

Las SVM pueden utilizarse en diferentes campos, incluyendo clasificación de documentos, reconocimiento de patrones, imágenes, análisis empresarial y procesamiento de datos científicos. También existe una variante denominada **SVR**, orientada a problemas de regresión.

Finalmente, las máquinas de vectores de soporte pueden integrarse dentro de procesos más amplios como **KDD, CRISP-DM y la Ruta de Ciencia de Datos**, especialmente durante las etapas de modelado y minería de datos.

En conclusión, SVM representa una técnica importante dentro del aprendizaje automático porque permite abordar problemas de clasificación utilizando principios matemáticos y geométricos para encontrar fronteras de decisión capaces de generalizar frente a nuevos datos.
