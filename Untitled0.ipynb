## 🔎 Modelos No Supervisados

* K-Means: Agrupa datos para segmentación de clientes.
* DBSCAN: Identifica zonas densas en análisis demográficos.
* Isolation Forest: Detecta anomalías para prevención de fraudes.

## 📊 Métricas de Evaluación

* Matriz de Confusión: Tabla de aciertos y errores.
* Precisión y Recall: Miden exactitud y exhaustividad.
* F1-Score: Equilibra la precisión y el recall.
* ROC y AUC: Evalúan el rendimiento del clasificador.

## 🗒️ Pipeline de Datos

* Imputación: Relleno de valores nulos o faltantes.
* Escalamiento: Normalización de escalas de variables numéricas.
* Reducción: Compresión dimensional para eliminar variables redundantes.
* Clasificador: Algoritmo que genera la predicción final.

## 📈 Optimización de Hiperparámetros

* GridSearchCV: Búsqueda exhaustiva, exacta pero muy costosa y lenta.
* RandomizedSearchCV: Exploración aleatoria, rápida y muy ágil.

## 🚀 Métodos de Ensamble

* Bagging: Entrena árboles en paralelo para reducir varianza.
* Boosting: Corrige errores de forma iterativa para reducir sesgo.

* Fuga de Datos: Error grave que invalida el modelo por mezclar test con entrenamiento.
