**Modelo predictivo para estimar precios de viviendas**
**Comparación entre Regresión Lineal y Árbol de Decisión**


En este proyecto voy a desarrollar un modelo predictivo que permita estimar el precio de una vivienda utilizando diferentes características de la propiedad.

Para realizar el proyecto utilizaré el Housing Prices Dataset disponible en Kaggle. El objetivo principal será comparar dos técnicas de aprendizaje automático: Regresión Lineal y Árbol de Decisión.

Voy a evaluar ambos modelos utilizando diferentes métricas de rendimiento, con el propósito de determinar cuál presenta un mejor comportamiento al momento de predecir los precios de las viviendas.

Durante todo el proyecto voy a explicar mediante comentarios personales qué hace cada instrucción y por qué considero necesario utilizarla dentro del proceso de construcción del modelo.

**Objetivo**

Construir y comparar dos modelos predictivos capaces de estimar el precio de las viviendas a partir de sus características, determinando cuál de los dos presenta un mejor rendimiento.

**Dataset**

El dataset utilizado corresponde a Housing Prices Dataset, disponible en Kaggle.

La variable que voy a predecir es price, que corresponde al precio de la vivienda.

Las demás variables representan características de las propiedades, como el área, número de habitaciones, baños, pisos, acceso a vía principal, sótano, aire acondicionado, parqueaderos y estado del amoblamiento.

**Exploración de los datos**

Antes de crear los modelos considero importante conocer el contenido del dataset.

Voy a revisar la cantidad de registros, las columnas disponibles, los tipos de datos y los posibles valores faltantes. Esto me permite entender con qué información estoy trabajando antes de comenzar el entrenamiento.

También voy a identificar cuáles variables son numéricas y cuáles son categóricas, ya que cada tipo de variable necesita un tratamiento diferente antes de utilizarla en un modelo predictivo.

**Preparación de los datos**

Para realizar la predicción voy a separar los datos en dos grupos.

El primer grupo será utilizado para entrenar los modelos. En este conjunto los algoritmos podrán aprender la relación entre las características de las viviendas y sus precios.

El segundo grupo será utilizado para realizar las pruebas. Estos datos no serán utilizados durante el entrenamiento, de manera que pueda comprobar si los modelos realmente son capaces de realizar predicciones sobre información que no conocen.

Utilizaré el 80 % de los datos para entrenamiento y el 20 % para prueba.

También será necesario transformar las variables categóricas a valores numéricos, ya que los modelos de aprendizaje automático necesitan trabajar con información numérica.

**Regresión Lineal**

El primer modelo que voy a utilizar será la Regresión Lineal.

Este modelo intenta encontrar una relación entre las características de las viviendas y su precio.

Lo utilizaré como modelo de referencia porque es relativamente sencillo y permite analizar si existe una relación aproximadamente lineal entre las características de las viviendas y su precio.

Después de entrenar el modelo voy a utilizarlo para realizar predicciones sobre el conjunto de prueba.

**Árbol de Decisión**

**El segundo modelo será un Árbol de Decisión.**

Este modelo funciona mediante diferentes reglas y divisiones de los datos. A diferencia de la regresión lineal, puede representar relaciones que no necesariamente sean lineales.

Voy a limitar la profundidad del árbol para evitar que el modelo se vuelva demasiado complejo y termine memorizando los datos de entrenamiento.

Después de entrenar el árbol utilizaré el conjunto de prueba para obtener sus predicciones.

**Evaluación de los modelos**

Para comparar los modelos voy a utilizar tres métricas principales.

**MAE: **representa el error absoluto promedio entre el precio real y el precio predicho. Un valor menor significa que el modelo está cometiendo errores más pequeños.

**RMSE: **representa la raíz del error cuadrático medio. Esta métrica es importante porque penaliza con mayor fuerza los errores grandes.

**R²:** permite conocer qué tan bien el modelo consigue explicar la variación existente en los precios. Un valor más cercano a 1 representa un mejor resultado.

**Comparación**

Después de realizar las predicciones voy a colocar los resultados de ambos modelos en una tabla comparativa.

La comparación tendrá en cuenta:

MAE.
RMSE.
R².

No voy a asumir previamente que la Regresión Lineal o el Árbol de Decisión será el mejor modelo.

La elección dependerá de los resultados obtenidos después de realizar las pruebas.

Si un modelo obtiene un MAE menor y un RMSE menor, significa que sus errores de predicción son menores.

Por otro lado, si obtiene un R² mayor, significa que consigue explicar una mayor proporción de la variación existente en los precios.

Por esta razón, considero que el mejor modelo será aquel que presente el equilibrio más favorable entre las diferentes métricas.

**Análisis**

La Regresión Lineal tiene como principal ventaja su sencillez. Es un modelo relativamente fácil de comprender y puede funcionar correctamente cuando existe una relación aproximadamente lineal entre las características y el precio.

El Árbol de Decisión es más flexible porque puede identificar diferentes reglas y relaciones entre las variables. Esto puede resultar útil cuando el precio de una vivienda depende de combinaciones de características que no siguen una relación lineal.

Sin embargo, tener un modelo más complejo no significa automáticamente obtener mejores resultados.

Por esta razón, considero importante realizar la evaluación utilizando datos que no hayan sido utilizados durante el entrenamiento.

Los resultados obtenidos mediante MAE, RMSE y R² permiten realizar una comparación objetiva entre los dos modelos.
