# MODELO PREDICTIVO PARA ESTIMAR EL PRECIO DE VIVIENDAS

## 1. Introducción

En este proyecto voy a desarrollar un modelo predictivo que permita estimar el precio de viviendas utilizando técnicas de aprendizaje automático. Para realizar el análisis voy a utilizar un conjunto de datos de viviendas que contiene diferentes características relacionadas con las propiedades.

El objetivo principal es comparar el rendimiento de dos modelos de Machine Learning: **Regresión Lineal** y **Árbol de Decisión**.

Para evaluar los modelos utilizaré diferentes métricas de rendimiento, principalmente el **Error Absoluto Medio (MAE)**, el **Error Cuadrático Medio (MSE)**, la **Raíz del Error Cuadrático Medio (RMSE)** y el **coeficiente de determinación R²**.

# 9. Identificar la variable objetivo

Para este proyecto la variable que quiero predecir es el **precio de la vivienda**.

En el dataset de viviendas de Kaggle utilizado como referencia, esta variable se denomina `SalePrice`.

```python
# Guardo la columna SalePrice como mi variable objetivo porque representa el precio que quiero predecir.
y = datos['SalePrice']
```

---

# 10. Seleccionar las variables predictoras

Para este proyecto voy a utilizar algunas características de las viviendas que pueden tener relación con su precio.

```python
# Selecciono las características de las viviendas que voy a utilizar para intentar explicar y predecir su precio.
caracteristicas = ['LotArea', 'YearBuilt', '1stFlrSF', '2ndFlrSF', 'FullBath', 'BedroomAbvGr', 'TotRmsAbvGrd']

# Creo el conjunto X solamente con las variables que seleccioné como características predictoras.
X = datos[caracteristicas].copy()
```

Las variables seleccionadas representan:

* `LotArea`: área del terreno.
* `YearBuilt`: año de construcción.
* `1stFlrSF`: superficie del primer piso.
* `2ndFlrSF`: superficie del segundo piso.
* `FullBath`: número de baños completos.
* `BedroomAbvGr`: número de habitaciones.
* `TotRmsAbvGrd`: cantidad total de habitaciones.

Estas variables fueron seleccionadas porque representan características físicas y estructurales de las viviendas que pueden tener relación con su precio.

# 12. Comprobar nuevamente los valores faltantes

```python
# Compruebo si existen precios faltantes porque necesito tener un valor objetivo para poder entrenar el modelo.
y.isnull().sum()
```

# 13. Eliminar registros incompletos

Si existen valores faltantes en las variables seleccionadas, puedo eliminar esos registros para simplificar el ejercicio.

```python
# Uno temporalmente las variables predictoras y el precio para poder eliminar de forma conjunta los registros incompletos.
datos_modelo = pd.concat([X, y], axis=1)

# Elimino las filas que tengan algún valor faltante porque necesito datos completos para entrenar los modelos.
datos_modelo = datos_modelo.dropna()

# Separo nuevamente las variables predictoras después de limpiar los datos.
X = datos_modelo[caracteristicas]

# Separo nuevamente la variable objetivo después de limpiar los datos.
y = datos_modelo['SalePrice']
```

---

# 14. Dividir los datos en entrenamiento y prueba

Voy a utilizar el **80 % de los datos para entrenamiento** y el **20 % para prueba**.

```python
# Divido mis datos para utilizar una parte en el aprendizaje y otra parte para comprobar el rendimiento de los modelos.
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

El `random_state=42` permite que la división sea reproducible, es decir, que al volver a ejecutar el código obtenga la misma separación.

---

# 15. Comprobar la división

```python
# Consulto cuántos registros quedaron para entrenar y comprobar el tamaño de mi conjunto de entrenamiento.
X_train.shape
```

```python
# Consulto cuántos registros quedaron para realizar las pruebas del modelo.
X_test.shape
```

---

# 16. MODELO 1: REGRESIÓN LINEAL

La regresión lineal intenta encontrar una relación matemática entre las variables de entrada y el precio de la vivienda.

## Crear el modelo

```python
# Creo el modelo de regresión lineal porque quiero utilizarlo como mi primer modelo predictivo.
modelo_lineal = LinearRegression()
```

## Entrenar el modelo

```python
# Entreno la regresión lineal utilizando las características y los precios conocidos del conjunto de entrenamiento.
modelo_lineal.fit(X_train, y_train)
```

## Realizar predicciones

```python
# Utilizo el modelo entrenado para predecir los precios de las viviendas que reservé para las pruebas.
predicciones_lineal = modelo_lineal.predict(X_test)
```

---

# 17. Evaluar la regresión lineal

## Error Absoluto Medio

```python
# Calculo el MAE para conocer en promedio cuánto se alejan mis predicciones del precio real.
mae_lineal = mean_absolute_error(y_test, predicciones_lineal)

# Muestro el resultado del error absoluto medio para evaluar el comportamiento de la regresión lineal.
print("MAE Regresión Lineal:", mae_lineal)
```

Un MAE menor significa que, en promedio, las predicciones están más cerca de los valores reales.

---

## Error Cuadrático Medio

```python
# Calculo el MSE porque quiero penalizar con mayor fuerza los errores grandes realizados por el modelo.
mse_lineal = mean_squared_error(y_test, predicciones_lineal)

# Muestro el resultado del error cuadrático medio de la regresión lineal.
print("MSE Regresión Lineal:", mse_lineal)
```

---

## Raíz del Error Cuadrático Medio

```python
# Calculo el RMSE porque quiero expresar el error nuevamente en las mismas unidades del precio de la vivienda.
rmse_lineal = np.sqrt(mse_lineal)

# Muestro el RMSE para tener una medida más fácil de interpretar.
print("RMSE Regresión Lineal:", rmse_lineal)
```

---

## Coeficiente R²

```python
# Calculo R2 para conocer qué proporción de la variación de los precios logra explicar mi modelo.
r2_lineal = r2_score(y_test, predicciones_lineal)

# Muestro el R2 para conocer la capacidad explicativa de la regresión lineal.
print("R² Regresión Lineal:", r2_lineal)
```

---

# 18. MODELO 2: ÁRBOL DE DECISIÓN

Ahora voy a construir un segundo modelo utilizando un **árbol de decisión para regresión**.

El árbol intenta dividir los datos en diferentes grupos tomando decisiones basadas en las características de las viviendas.

## Crear el modelo

```python
# Creo un árbol de decisión para comparar su capacidad predictiva con la regresión lineal.
modelo_arbol = DecisionTreeRegressor(random_state=42)
```

## Entrenar el árbol

```python
# Entreno el árbol de decisión utilizando las mismas variables y registros de entrenamiento que utilicé con la regresión lineal.
modelo_arbol.fit(X_train, y_train)
```

## Realizar predicciones

```python
# Utilizo el árbol entrenado para predecir los precios de las viviendas que se encuentran en el conjunto de prueba.
predicciones_arbol = modelo_arbol.predict(X_test)
```

---

# 19. Evaluar el árbol de decisión

## MAE

```python
# Calculo el MAE del árbol para saber qué tan alejadas están sus predicciones de los precios reales en promedio.
mae_arbol = mean_absolute_error(y_test, predicciones_arbol)

# Muestro el MAE obtenido por el árbol de decisión.
print("MAE Árbol de Decisión:", mae_arbol)
```

---

## MSE

```python
# Calculo el MSE del árbol para identificar el impacto de sus errores grandes.
mse_arbol = mean_squared_error(y_test, predicciones_arbol)

# Muestro el MSE obtenido por el árbol de decisión.
print("MSE Árbol de Decisión:", mse_arbol)
```

---

## RMSE

```python
# Calculo la raíz cuadrada del MSE para expresar nuevamente el error en unidades monetarias.
rmse_arbol = np.sqrt(mse_arbol)

# Muestro el RMSE obtenido por el árbol de decisión.
print("RMSE Árbol de Decisión:", rmse_arbol)
```

---

## R²

```python
# Calculo R2 para conocer qué tan bien logra explicar el árbol la variación existente en los precios.
r2_arbol = r2_score(y_test, predicciones_arbol)

# Muestro el R2 obtenido por el árbol de decisión.
print("R² Árbol de Decisión:", r2_arbol)
```

---

# 20. Comparación de los modelos

Ahora voy a organizar los resultados en una tabla para poder comparar los dos modelos.

```python
# Creo una tabla con las métricas obtenidas por ambos modelos para facilitar la comparación.
resultados = pd.DataFrame({
    'Modelo': ['Regresión Lineal', 'Árbol de Decisión'],
    'MAE': [mae_lineal, mae_arbol],
    'MSE': [mse_lineal, mse_arbol],
    'RMSE': [rmse_lineal, rmse_arbol],
    'R2': [r2_lineal, r2_arbol]
})

# Muestro la tabla para comparar directamente el rendimiento de los dos modelos.
resultados
```

---

# 21. Comparar el MAE

```python
# Creo una gráfica para comparar visualmente el error absoluto medio de los dos modelos.
plt.figure(figsize=(8, 5))

# Creo las barras utilizando los nombres de los modelos y sus respectivos valores de MAE.
plt.bar(resultados['Modelo'], resultados['MAE'])

# Coloco un título para explicar qué estoy comparando en la gráfica.
plt.title('Comparación del MAE entre modelos')

# Coloco el nombre del eje vertical porque representa el error absoluto medio.
plt.ylabel('MAE')

# Coloco el nombre del eje horizontal porque representa los modelos utilizados.
plt.xlabel('Modelo')

# Muestro la gráfica final.
plt.show()
```

### Interpretación

En esta gráfica debo observar cuál modelo presenta el **MAE más bajo**.

El modelo con menor MAE presenta un menor error promedio en sus predicciones.

---

# 22. Comparar el R²

```python
# Creo una gráfica para comparar el coeficiente R2 obtenido por los dos modelos.
plt.figure(figsize=(8, 5))

# Creo las barras con los valores de R2 de cada modelo.
plt.bar(resultados['Modelo'], resultados['R2'])

# Agrego un título para explicar el contenido de la gráfica.
plt.title('Comparación del R² entre modelos')

# Indico que el eje vertical representa el coeficiente R2.
plt.ylabel('R²')

# Indico que el eje horizontal contiene los modelos evaluados.
plt.xlabel('Modelo')

# Muestro la gráfica.
plt.show()
```

Un valor de R² más alto generalmente indica que el modelo explica una mayor proporción de la variabilidad observada en los precios.

---

# 23. Identificar automáticamente el mejor modelo

Para determinar el mejor modelo voy a utilizar principalmente el MAE, ya que permite interpretar directamente el error promedio de las predicciones.

```python
# Comparo los valores de MAE para identificar cuál modelo presenta el menor error.
if mae_lineal < mae_arbol:

    # Guardo el nombre de la regresión lineal porque obtuvo el menor MAE.
    mejor_modelo = "Regresión Lineal"

# Si la condición anterior no se cumple, significa que el árbol presentó menor error.
else:

    # Guardo el nombre del árbol de decisión porque obtuvo el menor MAE.
    mejor_modelo = "Árbol de Decisión"

# Muestro cuál de los dos modelos obtuvo el mejor resultado según el MAE.
print("El mejor modelo según el MAE es:", mejor_modelo)
```

# 26. Análisis de resultados

Después de entrenar y evaluar los dos modelos, puedo realizar el siguiente análisis.

La **Regresión Lineal** intenta representar la relación entre las características de las viviendas y su precio mediante una relación matemática. Su principal ventaja es que es un modelo relativamente sencillo de interpretar y entrenar.

El **Árbol de Decisión**, por otro lado, funciona mediante una estructura de decisiones que divide progresivamente los datos en grupos. Esto le permite representar relaciones que no necesariamente son lineales.

La comparación debe realizarse principalmente observando el **MAE**, **RMSE** y **R²**.

Para MAE y RMSE, un valor menor representa un menor error de predicción. En cambio, para R², un valor mayor generalmente representa una mejor capacidad del modelo para explicar la variación de los precios.

Es importante no elegir un modelo únicamente porque presenta un resultado favorable en una sola métrica. Lo recomendable es analizar todas las métricas y observar también el comportamiento de las predicciones.

# 27. Análisis personal

Desde mi análisis, considero que la comparación entre regresión lineal y árbol de decisión es importante porque ambos modelos intentan resolver el mismo problema utilizando estrategias diferentes.

La regresión lineal resulta interesante porque permite observar el problema desde una perspectiva matemática y asumir que existe una relación aproximadamente lineal entre las características de las viviendas y su precio.

El árbol de decisión puede representar relaciones más complejas, ya que puede dividir los datos dependiendo de diferentes condiciones. Esto puede resultar útil cuando el precio de una vivienda depende de combinaciones de características que no siguen una relación lineal sencilla.

Sin embargo, un árbol de decisión demasiado complejo puede aprender demasiado los datos de entrenamiento y presentar problemas de sobreajuste. Por esta razón, el hecho de que un modelo tenga un resultado excelente sobre los datos utilizados para entrenarlo no significa necesariamente que sea el mejor modelo para realizar predicciones sobre viviendas nuevas.

En este proyecto considero que el mejor modelo debe ser aquel que consiga un equilibrio entre un error bajo y una buena capacidad para generalizar sobre datos que no utilizó durante el entrenamiento.
