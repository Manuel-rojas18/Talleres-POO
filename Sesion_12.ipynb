# APRENDIZAJE POR CONJUNTOS (ENSEMBLE LEARNING)

## Introducción

En el aprendizaje automático existen numerosos algoritmos capaces de realizar predicciones a partir de datos. Algunos de ellos pueden producir buenos resultados de manera individual, pero en determinados problemas es posible mejorar el comportamiento de un sistema combinando varios modelos.

Esta idea da origen al **Aprendizaje por Conjuntos**, conocido en inglés como **Ensemble Learning**.

El aprendizaje por conjuntos consiste en utilizar varios modelos de aprendizaje automático y combinar sus resultados para obtener una predicción final. En lugar de depender exclusivamente de un único modelo, se construye un conjunto de modelos denominados **modelos base**, **estimadores** o **aprendices**, cuyas predicciones son posteriormente combinadas.

La idea fundamental puede representarse de la siguiente manera:

**Datos → varios modelos → combinación de resultados → predicción final**

Este enfoque es especialmente importante porque permite construir sistemas predictivos más robustos y constituye la base de algoritmos ampliamente utilizados como **Random Forest, AdaBoost y Gradient Boosting**.

---

# 1. ¿Qué es el aprendizaje por conjuntos?

El **aprendizaje por conjuntos** es una técnica de Machine Learning que combina las predicciones de varios modelos para producir una predicción final.

Cada modelo puede aprender diferentes patrones presentes en los datos.

En lugar de utilizar:

**Modelo único → Predicción**

se utiliza:

**Modelo 1 →**

**Modelo 2 →**

**Modelo 3 →** → **Combinación → Predicción final**

**Modelo 4 →**

La combinación puede realizarse mediante diferentes estrategias dependiendo del método utilizado.

---

# 2. ¿Por qué combinar varios modelos?

Un modelo individual puede cometer errores en determinados casos.

Si se utilizan varios modelos, es posible que los errores de unos sean compensados por las predicciones correctas de otros.

Por ejemplo, supongamos que tenemos tres modelos clasificadores:

* Modelo A → predice clase 1.
* Modelo B → predice clase 1.
* Modelo C → predice clase 2.

Si utilizamos votación mayoritaria, la predicción final sería:

**Clase 1**

La idea central consiste en aprovechar la información proporcionada por varios modelos.

---

# 3. Modelos base

Los modelos individuales que forman parte de un conjunto reciben diferentes nombres, entre ellos:

* Modelos base.
* Estimadores.

* Clasificadores base.

Pueden utilizarse diferentes algoritmos dependiendo del método.

Por ejemplo, un conjunto puede estar formado por:

* Árboles de decisión.
* Regresores.

* Máquinas de vectores de soporte.
* Modelos lineales.

Sin embargo, algunos métodos de Ensemble utilizan específicamente árboles de decisión como modelos base.

---

# 4. Principio de diversidad

Uno de los conceptos importantes del aprendizaje por conjuntos es la **diversidad**.

Si todos los modelos cometen exactamente los mismos errores, combinarlos no necesariamente proporciona una ventaja significativa.

Por el contrario, cuando los modelos aprenden diferentes patrones y cometen errores diferentes, la combinación puede resultar más útil.

De manera conceptual:

**Modelos diferentes → errores diferentes → combinación → mayor robustez**

Por esta razón, muchos algoritmos Ensemble buscan generar modelos que no sean completamente idénticos entre sí.

# 6. Bagging

**Bagging** significa **Bootstrap Aggregating**.

La idea consiste en crear diferentes conjuntos de entrenamiento a partir del conjunto de datos original utilizando muestreo con reemplazo.

Después se entrena un modelo independiente con cada conjunto.

Finalmente, se combinan sus predicciones.

El proceso puede representarse así:

**Datos originales**

↓

**Muestra 1 → Modelo 1**

**Muestra 2 → Modelo 2**

**Muestra 3 → Modelo 3**

**Muestra 4 → Modelo 4**

↓

**Agregación**

↓

**Predicción final**

---

# 7. ¿Qué es Bootstrap?

Bootstrap es una técnica de muestreo que permite crear nuevos conjuntos de datos a partir de un conjunto original seleccionando observaciones con reemplazo.

Por ejemplo, si tenemos:

**A B C D E**

podemos crear una muestra como:

**A C C E B**

Otra podría ser:

**D A B B E**

Al permitir que un elemento aparezca más de una vez, se generan diferentes conjuntos de entrenamiento.

Esto ayuda a producir modelos diferentes.

---

# 8. Agregación

Después de entrenar los diferentes modelos, se combinan sus predicciones.

En clasificación puede utilizarse la **votación mayoritaria**.

Por ejemplo:

| Modelo   | Predicción |
| -------- | ---------- |
| Modelo 1 | A          |
| Modelo 2 | B          |
| Modelo 3 | A          |
| Modelo 4 | A          |
| Modelo 5 | B          |

La clase A recibe tres votos y la clase B recibe dos.

Por lo tanto:

**Predicción final = A**

En regresión, normalmente puede utilizarse un promedio de las predicciones.

---

# 9. Random Forest

Uno de los algoritmos Ensemble más conocidos es **Random Forest**.

Random Forest significa **Bosque Aleatorio**.

Su funcionamiento se basa en construir múltiples árboles de decisión y combinar sus resultados.

La idea puede representarse así:

**Datos**

↓

**Árbol 1**

**Árbol 2**

**Árbol 3**

**Árbol 4**

**Árbol 5**

↓

**Combinación**

↓

**Predicción final**

El objetivo es obtener un modelo más robusto que un único árbol de decisión.

---

# 10. ¿Por qué se llama Random Forest?

El término **Forest**, o bosque, hace referencia a que existen muchos árboles.

La palabra **Random**, o aleatorio, se relaciona con el uso de elementos aleatorios durante la construcción de los árboles.

Esto permite que los diferentes árboles no sean exactamente iguales.

Cada árbol puede observar diferentes muestras de datos y diferentes subconjuntos de características.

---

# 11. Random Forest para clasificación

En clasificación, cada árbol realiza una predicción.

Por ejemplo:

| Árbol   | Predicción |
| ------- | ---------- |
| Árbol 1 | Clase A    |
| Árbol 2 | Clase B    |
| Árbol 3 | Clase A    |
| Árbol 4 | Clase A    |
| Árbol 5 | Clase B    |

La clase A obtiene tres votos y la clase B obtiene dos.

La predicción final sería:

**Clase A**

Este mecanismo se conoce como **votación mayoritaria**.

---

# 12. Random Forest para regresión

Por ejemplo, si queremos predecir el precio de una vivienda:

| Árbol   | Predicción |

| Árbol 1 |   $240.000 |
| Árbol 2 |   $250.000 |
| Árbol 3 |   $245.000 |
| Árbol 4 |   $260.000 |
| Árbol 5 |   $255.000 |

El modelo puede combinar estas predicciones utilizando un promedio.

El resultado sería una estimación conjunta del precio.

# 14. Boosting

Otra estrategia importante de aprendizaje por conjuntos es **Boosting**.

A diferencia de Bagging, donde los modelos suelen entrenarse de manera independiente, Boosting construye los modelos de forma secuencial.

La idea básica es:

**Modelo 1**

↓

**Identificar errores**

↓

**Modelo 2**

↓

**Identificar errores**

↓

**Modelo 3**

↓

**Combinar modelos**

↓

**Predicción final**

Cada nuevo modelo intenta mejorar determinados aspectos del conjunto construido anteriormente.

---

# 15. AdaBoost

**AdaBoost**, cuyo nombre proviene de **Adaptive Boosting**, es uno de los algoritmos clásicos de Boosting.

Su idea consiste en prestar mayor atención a los registros que los modelos anteriores clasificaron incorrectamente.

De esta forma, los siguientes modelos pueden concentrarse más en los casos difíciles.

El proceso puede representarse como:

**Modelo 1 → errores**

↓

**Aumentar atención a errores**

↓

**Modelo 2**

↓

**Nuevos errores**

↓

**Modelo 3**

↓

**Combinación**

---

# 16. Gradient Boosting

**Gradient Boosting** es otra técnica de aprendizaje por conjuntos.

En este caso, los modelos se construyen de manera secuencial y cada nuevo modelo intenta corregir los errores o residuos producidos por el conjunto anterior.

En regresión puede explicarse de manera conceptual:

**Predicción inicial**

↓

**Calcular errores**

↓

**Aprender los errores**

↓

**Actualizar la predicción**

↓

**Calcular nuevos errores**

↓

**Repetir**

Cada nuevo modelo contribuye a mejorar la predicción final.

---

# 17. Idea matemática simplificada de Gradient Boosting

Supongamos que tenemos una predicción inicial:

**F₀(x)**

El siguiente modelo aprende una parte del error:

**h₁(x)**

La nueva predicción puede representarse conceptualmente como:

**F₁(x) = F₀(x) + ηh₁(x)**

Donde:

* **F₀(x)** representa el modelo anterior.
* **h₁(x)** representa el nuevo modelo.
* **η** representa la tasa de aprendizaje.

Este proceso continúa durante varias iteraciones.

---

# 18. Tasa de aprendizaje

La **learning rate**, o tasa de aprendizaje, controla cuánto contribuye cada nuevo modelo a la predicción final.

Una tasa de aprendizaje pequeña hace que cada modelo contribuya gradualmente.

Una tasa mayor permite actualizaciones más grandes.

Por ejemplo:

```python id="wz8d2f"
GradientBoostingRegressor(
    learning_rate=0.05
)
```

El valor exacto debe seleccionarse mediante experimentación y evaluación.

---

# 19. Número de estimadores

Otro hiperparámetro importante es el número de modelos que se utilizan.

Por ejemplo:

```python id="r8h1ka"
GradientBoostingRegressor(
    n_estimators=100
)
```

En este caso se construirían 100 estimadores durante el proceso.

Existe una relación entre:

* Número de estimadores.
* Tasa de aprendizaje.
* Complejidad.
* Tiempo de entrenamiento.

Estos elementos deben analizarse conjuntamente.

# 20. Stacking

Otra estrategia de Ensemble es **Stacking**.

En lugar de combinar directamente las predicciones mediante una votación o promedio, Stacking utiliza un modelo adicional que aprende cómo combinar las predicciones de otros modelos.

Por ejemplo:

**Regresión Lineal**

**Árbol de Decisión**

**SVM**

↓

**Predicciones**

↓

**Modelo combinador**

↓

**Predicción final**

El modelo que combina los resultados recibe el nombre de **meta-modelo**.

---

# 21. Voting

Voting es una técnica que combina las predicciones de diferentes modelos.

En clasificación puede utilizarse:

**Hard Voting**

donde se selecciona la clase que obtiene más votos.

También existe:

**Soft Voting**

donde se combinan las probabilidades proporcionadas por los diferentes modelos.

Por ejemplo:

| Modelo   | Clase A | Clase B |
| -------- | ------: | ------: |
| Modelo 1 |    0.80 |    0.20 |
| Modelo 2 |    0.70 |    0.30 |
| Modelo 3 |    0.60 |    0.40 |

El conjunto puede combinar estas probabilidades para obtener una predicción final.

---

# 22. Diferencia entre Bagging y Boosting

Una diferencia fundamental es la forma en que se entrenan los modelos.

| Característica            | Bagging                    | Boosting                        |
| ------------------------- | -------------------------- | ------------------------------- |
| Entrenamiento             | Generalmente independiente | Secuencial                      |
| Objetivo                  | Reducir variabilidad       | Mejorar errores progresivamente |
| Ejemplo                   | Random Forest              | AdaBoost                        |
| Dependencia entre modelos | Menor                      | Mayor                           |
| Enfoque                   | Combinar modelos           | Corregir errores                |

Esta diferencia permite comprender por qué ambos métodos pertenecen al aprendizaje por conjuntos, pero utilizan estrategias diferentes.

---

# 23. Árboles de decisión como modelos base

Los árboles de decisión son utilizados frecuentemente como modelos base en Ensemble Learning.

Esto ocurre porque un árbol individual puede tener limitaciones.

Por ejemplo, un árbol muy profundo puede sobreajustar los datos.

En un Ensemble se pueden combinar muchos árboles para obtener un comportamiento diferente.

Por esta razón:

**Árbol de decisión → modelo individual**

**Random Forest → conjunto de árboles**

**Gradient Boosting → conjunto secuencial de modelos**

---

# 24. Reducción del sobreajuste

Una de las motivaciones para utilizar Ensemble Learning es reducir determinados problemas de los modelos individuales.

En el caso de Bagging y Random Forest, combinar diferentes modelos puede ayudar a reducir la variabilidad de las predicciones.

Sin embargo, esto no significa que un Ensemble sea automáticamente inmune al sobreajuste.

Un modelo demasiado complejo o una configuración inadecuada todavía puede presentar problemas de generalización.

Por esta razón siempre es necesario evaluar el modelo con datos apropiados.

---

# 25. Sesgo y varianza

En Machine Learning se suelen estudiar dos conceptos importantes:

* **Sesgo (Bias)**
* **Varianza (Variance)**

Un modelo con mucho sesgo puede ser demasiado simple.

Un modelo con mucha varianza puede ser demasiado sensible a los datos de entrenamiento.

Los métodos Ensemble pueden ayudar a modificar este equilibrio dependiendo de la estrategia utilizada.

De manera general:

**Bagging → puede ayudar a reducir varianza**

**Boosting → puede ayudar a reducir sesgo mediante modelos sucesivos**

Esta es una simplificación conceptual; el comportamiento real depende del algoritmo y de los datos.

# Conclusión

El **Aprendizaje por Conjuntos o Ensemble Learning** es una estrategia de aprendizaje automático que combina diferentes modelos para producir una predicción final.

Su importancia se encuentra en la posibilidad de aprovechar las capacidades de múltiples modelos y, dependiendo del método y de los datos, obtener sistemas más robustos y con mejor capacidad de generalización.

Entre las principales estrategias se encuentran **Bagging, Random Forest, Boosting, AdaBoost, Gradient Boosting, Voting y Stacking**. Cada una presenta una forma diferente de construir y combinar los modelos.

**Bagging** trabaja generalmente mediante modelos independientes entrenados sobre diferentes muestras de los datos. **Random Forest** lleva esta idea al utilizar múltiples árboles de decisión. **Boosting**, por otro lado, construye modelos secuencialmente, intentando mejorar los errores de los modelos anteriores. **Gradient Boosting** representa una de las técnicas más importantes de esta familia y utiliza modelos sucesivos para reducir progresivamente el error.

El aprendizaje por conjuntos también permite comprender conceptos fundamentales como diversidad de modelos, sesgo, varianza, sobreajuste, validación cruzada e hiperparámetros.

En el contexto de los temas estudiados anteriormente, Ensemble Learning se relaciona directamente con los **árboles de decisión**, ya que estos pueden utilizarse como modelos base; con **KDD**, porque puede participar en la etapa de minería de datos; con **CRISP-DM**, porque puede utilizarse durante la fase de modelado; y con la **Ruta de Ciencia de Datos**, porque representa una alternativa para construir modelos predictivos.

En un proyecto como la predicción del precio de viviendas, por ejemplo, es posible comparar una regresión lineal, un árbol de decisión, Random Forest y Gradient Boosting utilizando métricas como MAE, RMSE y R². De esta manera, se puede estudiar experimentalmente cómo diferentes estrategias de modelado responden ante un mismo conjunto de datos.

En conclusión, el aprendizaje por conjuntos constituye una herramienta fundamental dentro del Machine Learning porque permite pasar de la utilización de modelos individuales a sistemas que combinan múltiples aprendices. Su estudio permite comprender mejor cómo se construyen modelos predictivos modernos y cómo diferentes estrategias pueden utilizarse para analizar datos, descubrir patrones y realizar predicciones.
