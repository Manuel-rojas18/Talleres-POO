# DEL CAOS AL ORDEN: APRENDIZAJE NO SUPERVISADO Y DESCUBRIMIENTO DE PATRONES

## Introducción

En los proyectos de Ciencia de Datos es común encontrar grandes cantidades de información que inicialmente parecen desordenadas, difíciles de interpretar y sin una clasificación previamente establecida. Los datos pueden provenir de clientes, sensores, transacciones, imágenes, registros académicos, sistemas empresariales o diferentes fuentes digitales.

En muchas ocasiones conocemos el resultado que queremos predecir y contamos con una variable objetivo. Sin embargo, también existen situaciones en las que tenemos los datos, pero **no conocemos previamente las categorías, grupos o patrones que contienen**.

En este contexto aparece el **aprendizaje no supervisado**, una rama del Machine Learning que busca descubrir estructuras, agrupaciones y comportamientos dentro de los datos sin utilizar una variable objetivo previamente etiquetada.

Por esta razón, podemos interpretar el aprendizaje no supervisado como un proceso de:

**Caos → análisis → descubrimiento de patrones → organización → conocimiento**

La idea de **“Del caos al orden”** representa precisamente este proceso. Partimos de un conjunto de datos aparentemente desorganizado y utilizamos algoritmos para encontrar estructuras que no habían sido definidas previamente.

Entre las técnicas más importantes encontramos:

1. Datos no supervisados.
2. Agrupamiento y K-Means.
3. Segmentación.
4. DBSCAN y detección de formas.
5. Anomalías y novedades.
6. Mezclas gaussianas.

---

# 1. ¿Qué es el aprendizaje no supervisado?

El **aprendizaje no supervisado** es una categoría del aprendizaje automático en la que los algoritmos trabajan con datos que no tienen una etiqueta o variable objetivo conocida.

En aprendizaje supervisado tenemos:

**Datos + respuestas conocidas → entrenamiento → predicción**

Mientras que en aprendizaje no supervisado tenemos:

**Datos sin etiquetas → algoritmo → descubrimiento de patrones**

Por ejemplo, imaginemos una empresa que tiene información sobre miles de clientes:

* Edad.
* Ingresos.
* Número de compras.
* Frecuencia de compra.
* Valor promedio de las compras.

Si la empresa no ha clasificado previamente a los clientes, un algoritmo no supervisado puede intentar descubrir grupos con características similares.

El algoritmo podría encontrar, por ejemplo, diferentes grupos de comportamiento sin que nosotros le indiquemos previamente cuáles son.

# 2. Datos no supervisados

Los **datos no supervisados** son datos en los que no existe una variable objetivo o etiqueta que indique explícitamente la categoría de cada observación.

| ------- | ---: | ------: | --------: |

No tenemos una columna llamada:

**“Tipo de cliente”**

El algoritmo debe analizar los datos y descubrir si existen grupos naturales.

La diferencia principal está en la existencia de etiquetas.

| Característica    | Supervisado               | No supervisado               |

Por ejemplo, si queremos predecir el precio de una vivienda utilizando `SalePrice`, estamos ante un problema supervisado.

Si queremos descubrir grupos de viviendas similares sin tener categorías previamente definidas, podemos utilizar aprendizaje no supervisado.

# 5. Agrupamiento o Clustering

El **clustering** es una de las principales técnicas de aprendizaje no supervisado.

Consiste en agrupar observaciones que presentan características similares.

La idea general es:

**Datos**

↓

**Algoritmo de agrupamiento**

↓

**Grupo 1**

**Grupo 2**

**Grupo 3**

**Grupo 4**

Cada grupo recibe el nombre de **cluster**.

---

# 6. ¿Qué significa que dos datos sean similares?

Para agrupar datos necesitamos establecer alguna forma de medir la similitud o distancia.

Una de las medidas más conocidas es la **distancia euclidiana**.

Para dos puntos en un espacio de dos dimensiones:

**A(x₁, y₁)**

y

**B(x₂, y₂)**

la distancia puede calcularse como:

**d = √((x₂ − x₁)² + (y₂ − y₁)²)**

Cuanto menor sea la distancia, mayor puede ser la similitud entre los puntos según esta medida.

# 7. K-Means

Uno de los algoritmos de agrupamiento más conocidos es **K-Means**.

Su objetivo es dividir los datos en un número determinado de grupos o clusters.

La letra **K** representa el número de grupos que queremos construir.

Por ejemplo:

**K = 3**

significa que queremos obtener tres clusters.

---

# 8. Funcionamiento de K-Means

El funcionamiento básico de K-Means puede resumirse de la siguiente manera:

### Paso 1: seleccionar K

Se define el número de grupos.

Por ejemplo:

**K = 3**

### Paso 2: seleccionar centroides iniciales

El algoritmo establece posiciones iniciales para los centroides.

### Paso 3: asignar puntos

Cada observación se asigna al centroide más cercano.

### Paso 4: recalcular centroides

Se calcula una nueva posición para cada centroide.

### Paso 5: repetir

Se vuelven a asignar los puntos y actualizar los centroides.

### Paso 6: detenerse

El proceso continúa hasta que las posiciones de los centroides se estabilizan o se alcanza el número máximo de iteraciones.

---

# 9. Los centroides

Un **centroide** representa el centro de un cluster.

Podemos imaginarlo como el punto que representa la posición promedio de los elementos pertenecientes a un grupo.

Por ejemplo:

```text
        ● ●
      ● ● X ●
        ● ●

              X

                    ●
                  ● ●
```

Las `X` representan posibles centroides.

Los puntos cercanos a cada centroide forman parte del cluster correspondiente.

---

# 10. Ejemplo de K-Means

Supongamos que tenemos clientes representados por:

* Ingresos.
* Número de compras.

Podemos solicitar:

**K = 3**

El algoritmo podría encontrar:

### Cluster 1

Clientes con pocos ingresos y pocas compras.

### Cluster 2

Clientes con ingresos medios y compras moderadas.

### Cluster 3

Clientes con ingresos altos y muchas compras.

Es importante recordar que el algoritmo no conoce previamente estos nombres. Los grupos son descubiertos a partir de los patrones presentes en los datos.

---

# 11. Selección del valor K

Uno de los principales desafíos de K-Means consiste en determinar cuántos clusters utilizar.

Una técnica frecuente es el **método del codo**, conocido como **Elbow Method**.

Se ejecuta K-Means con diferentes valores de K y se analiza cómo cambia la suma de errores dentro de los clusters.

Por ejemplo:

**K = 2**

**K = 3**

**K = 4**

**K = 5**

**K = 6**

Se busca un punto donde aumentar el número de clusters deje de producir mejoras importantes.

---

# 12. Inercia

En K-Means se utiliza frecuentemente la **inercia** para evaluar qué tan cerca están los puntos de sus centroides.

Una representación simplificada sería:

**Inercia = suma de las distancias cuadráticas entre cada punto y su centroide**

Una menor inercia significa que los puntos están más próximos a sus centroides.

Sin embargo, aumentar K normalmente reduce la inercia, por lo que no se debe elegir simplemente el K que produzca la menor inercia.

# 16. DBSCAN

Otro algoritmo importante de aprendizaje no supervisado es **DBSCAN**.

Su nombre significa:

**Density-Based Spatial Clustering of Applications with Noise**

Es un algoritmo de agrupamiento basado en densidad.

A diferencia de K-Means, DBSCAN no necesita que indiquemos directamente el número de clusters.

Esto representa una diferencia importante.

---

# 17. ¿Cómo funciona DBSCAN?

DBSCAN busca regiones donde existen muchos puntos cercanos.

En términos sencillos:

**Alta densidad de puntos → posible cluster**

**Baja densidad → posible separación o ruido**

Esto permite encontrar grupos con formas que no necesariamente son circulares.

---

# 18. Puntos de ruido

Una característica importante de DBSCAN es que puede identificar puntos que no pertenecen claramente a ningún cluster.

Estos puntos pueden considerarse:

**Ruido**

o

**Outliers**

Esto es especialmente útil cuando queremos encontrar agrupaciones y detectar observaciones alejadas del comportamiento general.

---

# 19. Parámetros principales de DBSCAN

DBSCAN utiliza principalmente dos parámetros:

### `eps`

Define una distancia máxima para considerar que dos puntos están suficientemente cerca.

### `min_samples`

Indica el número mínimo de puntos necesario para considerar que existe una región suficientemente densa.

Por ejemplo:

```python id="m4n5ps"
from sklearn.cluster import DBSCAN

modelo = DBSCAN(
    eps=0.5,
    min_samples=5
)
```

Los valores deben ajustarse según las características del conjunto de datos.

---

# 20. DBSCAN y las formas

Una ventaja importante de DBSCAN es que puede encontrar clusters con formas más complejas.

Por ejemplo, imaginemos dos grupos con forma de media luna.

K-Means puede presentar dificultades porque sus clusters están relacionados con la distancia a los centroides.

DBSCAN, al trabajar con densidad, puede identificar estructuras más complejas.

Por eso en la imagen aparece el concepto:

**“DBSCAN y Formas”**

La técnica resulta especialmente interesante cuando los grupos no presentan formas simples.

---

# 21. Comparación entre K-Means y DBSCAN

| Característica         | K-Means                     | DBSCAN                    |
| ---------------------- | --------------------------- | ------------------------- |
| Número de clusters     | Se debe definir K           | No se define directamente |
| Basado en              | Centroides                  | Densidad                  |
| Formas complejas       | Puede tener dificultades    | Puede manejarlas          |
| Detección de ruido     | No es su objetivo principal | Sí                        |
| Parámetros principales | K                           | `eps`, `min_samples`      |

Cada algoritmo tiene situaciones donde puede resultar más apropiado.

---

# 22. Anomalías y novedades

El aprendizaje no supervisado también puede utilizarse para detectar comportamientos diferentes.

Una **anomalía** es una observación que presenta características poco comunes respecto al comportamiento general de los datos.

Por ejemplo:

En un sistema de transacciones:

* La mayoría de compras son pequeñas.
* Una transacción extremadamente diferente puede requerir análisis.

La detección de anomalías puede ayudar a identificar situaciones inusuales.

---

# 23. Anomalías

Una anomalía puede ser:

* Un dato atípico.
* Un comportamiento poco frecuente.
* Un registro incorrecto.
* Una situación excepcional.
* Un evento que merece investigación.

Es importante aclarar que:

**Anomalía no significa automáticamente error.**

Una observación puede ser poco común y aun así ser correcta.

---

# 24. Novedades

El concepto de **novedad** está relacionado con identificar observaciones nuevas que no corresponden al patrón conocido.

Por ejemplo, un sistema puede aprender el comportamiento normal de una máquina.

Cuando aparece un nuevo registro con características diferentes, el sistema puede marcarlo como potencial novedad.

Esto puede utilizarse en:

* Sistemas industriales.
* Seguridad informática.
* Monitoreo de sensores.
* Finanzas.
* Sistemas de detección.

---

# 25. Diferencia entre anomalía y novedad

Aunque están relacionadas, pueden tener enfoques diferentes.

### Detección de anomalías

Busca identificar observaciones inusuales dentro de un conjunto de datos.

### Detección de novedades

Busca determinar si nuevas observaciones se alejan del comportamiento previamente conocido.

En ambos casos se intenta identificar comportamientos que requieren atención.

---

# 26. Mezclas gaussianas

Otra técnica importante dentro del aprendizaje no supervisado son las **Mezclas Gaussianas**, conocidas como **Gaussian Mixture Models o GMM**.

Este método representa los datos como una combinación de diferentes distribuciones gaussianas.

En lugar de asumir que cada grupo está representado simplemente por un centro, GMM considera una distribución probabilística.

---

# 27. ¿Qué es una distribución gaussiana?

Una distribución gaussiana, también conocida como distribución normal, tiene una forma característica similar a una campana.

Conceptualmente:

```text
             *
           *   *
         *       *
       *           *
_____*_______________*____
```

La distribución está relacionada con conceptos como:

* Media.
* Varianza.
* Desviación estándar.

---

# 28. ¿Cómo funcionan las mezclas gaussianas?

Una mezcla gaussiana supone que los datos pueden ser generados por diferentes distribuciones gaussianas.

Por ejemplo:

**Datos**

↓

**Gaussiana 1**

**Gaussiana 2**

**Gaussiana 3**

↓

**Combinación**

↓

**Representación de los datos**

Cada distribución puede representar una posible estructura dentro del conjunto de datos.

---

# 29. K-Means frente a Gaussian Mixture Models

K-Means asigna cada punto a un cluster determinado.

Por ejemplo:

**Punto → Cluster 1**

Mientras que un modelo de mezcla gaussiana puede proporcionar probabilidades de pertenencia.

Por ejemplo:

**Punto → 70% Cluster 1**

**Punto → 25% Cluster 2**

**Punto → 5% Cluster 3**

Esto se conoce como una asignación más **probabilística o suave**.

---

# 30. Clustering duro y clustering suave

Podemos diferenciar:

### Clustering duro

Cada punto pertenece a un único grupo.

Ejemplo:

**Cliente → Cluster 2**

### Clustering suave

Cada punto puede tener diferentes probabilidades de pertenecer a varios grupos.

Ejemplo:

**Cliente → 60% Cluster 1**

**Cliente → 35% Cluster 2**

**Cliente → 5% Cluster 3**

Las mezclas gaussianas permiten trabajar con este segundo enfoque.

---

# 31. Algoritmo EM

Las mezclas gaussianas suelen estimarse utilizando el algoritmo **Expectation-Maximization**, conocido como **EM**.

El proceso se puede explicar en dos etapas principales:

### Expectation

Se estiman las probabilidades de pertenencia de los puntos a las diferentes distribuciones.

### Maximization

Se actualizan los parámetros de las distribuciones utilizando esas probabilidades.

Este proceso se repite hasta alcanzar una condición de convergencia.

---

# 42. Evaluación del clustering

A diferencia del aprendizaje supervisado, en clustering normalmente no tenemos una etiqueta conocida para decir inmediatamente si la predicción es correcta.

Por eso pueden utilizarse diferentes métricas.

Una de ellas es el:

**Silhouette Score**

El coeficiente de silueta permite analizar qué tan similares son los puntos respecto a su propio cluster y qué tan separados están de otros clusters.

También pueden utilizarse:

* Inercia.
* Davies-Bouldin Index.
* Calinski-Harabasz Index.

Estas métricas deben interpretarse de acuerdo con el problema.

---

# 43. Silhouette Score

El Silhouette Score tiene valores entre:

**-1 y 1**

De manera general:

* Valores cercanos a 1 → buena separación.
* Valores cercanos a 0 → clusters cercanos o poco definidos.
* Valores negativos → posibles asignaciones problemáticas.

No debe utilizarse como único criterio para decidir si una agrupación tiene sentido.

También es necesario considerar el contexto de los datos.

# 51. Importancia del conocimiento del dominio

Un algoritmo puede encontrar grupos matemáticamente válidos, pero esto no significa que automáticamente tengan significado para una organización.

Por ejemplo, K-Means puede crear tres clusters de clientes.

El algoritmo solamente sabe que existen diferencias matemáticas entre los grupos.

Un analista debe estudiar:

* Qué características diferencian los grupos.
* Si los grupos tienen sentido.
* Qué comportamiento representa cada uno.
* Qué decisiones pueden apoyarse en estos resultados.

Por esta razón, Machine Learning y conocimiento del dominio deben trabajar conjuntamente.

# Conclusión

El **aprendizaje no supervisado** representa una de las áreas fundamentales del Machine Learning porque permite analizar datos cuando no contamos con etiquetas o categorías previamente definidas.

El concepto de **“Del caos al orden”** permite comprender de manera sencilla el propósito de estas técnicas. Partimos de un conjunto de datos que puede parecer desorganizado y utilizamos algoritmos matemáticos para descubrir estructuras, agrupaciones, comportamientos y patrones.

Dentro de este proceso, el **agrupamiento y K-Means** permiten dividir los datos en grupos utilizando centroides. La **segmentación** permite interpretar esos grupos y encontrar conjuntos de observaciones con características similares. **DBSCAN** introduce un enfoque basado en densidad que permite encontrar estructuras con formas más complejas y detectar puntos considerados como ruido.

Por otra parte, el análisis de **anomalías y novedades** permite identificar comportamientos que se alejan de los patrones habituales. Esto puede ser útil en diferentes áreas como industria, finanzas, ciberseguridad y monitoreo de sistemas.

Las **mezclas gaussianas** proporcionan una perspectiva probabilística del agrupamiento, permitiendo representar la pertenencia de los datos a diferentes grupos mediante probabilidades. Esto las diferencia de métodos como K-Means, donde normalmente cada observación queda asignada a un cluster específico.

Estas técnicas no funcionan de manera aislada. Forman parte de procesos completos de Ciencia de Datos y pueden relacionarse con **KDD, CRISP-DM, reducción de dimensionalidad y exploración de datos**.

El aprendizaje no supervisado demuestra que los datos pueden contener estructuras que no son evidentes a simple vista. Mediante técnicas como K-Means, DBSCAN, detección de anomalías y Gaussian Mixture Models es posible transformar grandes cantidades de información aparentemente desordenada en representaciones más comprensibles.

Por ello, **“Del caos al orden”** no significa simplemente organizar datos, sino utilizar métodos matemáticos y computacionales para descubrir patrones ocultos, interpretarlos y convertirlos en conocimiento útil para la toma de decisiones.

En última instancia, el verdadero objetivo del aprendizaje no supervisado no es solamente encontrar grupos, sino **comprender qué significan esos grupos y qué información nueva pueden aportar sobre el fenómeno que estamos estudiando**.
